```python
data = {
    "Nama": ["Adi", "Budi", "Cinny", "Delta", "Eki",
             "Fara", "Gandhi", "Hadi", "Ijah", "Jerome",
             "Kirana", "Lopan", "Mina"],

    "NIM": [110, 220, 330, 440, 550, 660, 770,
            880, 990, 111, 221, 331, 441],

    "Gender": ["M", "M", "F", "F", "M", "F", "M",
               "M", "F", "M", "F", "M", "F"],

    "Study Hours": [10, 19, 4, 1, 9, 3, 21,
                    3, 13, 15, 7, 0, 11],

    "Exam Score": [82, 95, 70, 45, 85, 60, 92,
                   67, 81, 97, 60, 30, 86]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(pd.DataFrame(data))
```

```python
df["Exam Result"] = df["Exam Score"].apply(
    lambda x: "Pass" if x > 65 else "Fail"
)

print(df)
```

```python
df["Exam Result"] = ""

for i in range(len(df)):
    if df.loc[i, "Exam Score"] > 65:
        df.loc[i, "Exam Result"] = "Pass"
    else:
        df.loc[i, "Exam Result"] = "Fail"

print(df)
```

```python
avg_score = df.groupby("Gender")["Exam Score"].mean()

print(avg_score)
```

```python
avg_score.plot(kind="bar")

plt.xlabel("Gender")
plt.ylabel("Average Exam Score")
plt.title("Average Exam Score by Gender")
plt.show()
```

```python
avg_score = df.groupby("Gender")["Exam Score"].mean()
avg_score.plot(kind="bar")
plt.xlabel("Gender")
plt.ylabel("Average Exam Score")
plt.title("Average Exam Score by Gender")
plt.show()
```
